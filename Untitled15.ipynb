```python
%%writefile /content/model.py

import torch
import torch.nn as nn


class Generator(nn.Module):
    def __init__(self):
        super().__init__()

        self.net = nn.Sequential(
            nn.Conv1d(1, 32, kernel_size=15, padding=7),
            nn.LeakyReLU(0.2),

            nn.Conv1d(32, 64, kernel_size=15, stride=2, padding=7),
            nn.BatchNorm1d(64),
            nn.LeakyReLU(0.2),

            nn.Conv1d(64, 128, kernel_size=15, stride=2, padding=7),
            nn.BatchNorm1d(128),
            nn.LeakyReLU(0.2),

            nn.ConvTranspose1d(
                128, 64, kernel_size=4, stride=2, padding=1
            ),
            nn.BatchNorm1d(64),
            nn.ReLU(),

            nn.ConvTranspose1d(
                64, 32, kernel_size=4, stride=2, padding=1
            ),
            nn.BatchNorm1d(32),
            nn.ReLU(),

            nn.Conv1d(32, 1, kernel_size=15, padding=7),
            nn.Tanh()
        )

    def forward(self, noisy_audio):
        enhanced = self.net(noisy_audio)

        # Ensure the output length matches the input length.
        if enhanced.shape[-1] > noisy_audio.shape[-1]:
            enhanced = enhanced[..., :noisy_audio.shape[-1]]
        elif enhanced.shape[-1] < noisy_audio.shape[-1]:
            enhanced = nn.functional.pad(
                enhanced,
                (0, noisy_audio.shape[-1] - enhanced.shape[-1])
            )

        return enhanced


class Discriminator(nn.Module):
    def __init__(self):
        super().__init__()

        self.net = nn.Sequential(
            nn.Conv1d(2, 32, kernel_size=15, stride=2, padding=7),
            nn.LeakyReLU(0.2),

            nn.Conv1d(32, 64, kernel_size=15, stride=2, padding=7),
            nn.BatchNorm1d(64),
            nn.LeakyReLU(0.2),

            nn.Conv1d(64, 128, kernel_size=15, stride=2, padding=7),
            nn.BatchNorm1d(128),
            nn.LeakyReLU(0.2),

            nn.Conv1d(128, 1, kernel_size=15, padding=7)
        )

    def forward(self, noisy_audio, clean_audio):
        combined = torch.cat([noisy_audio, clean_audio], dim=1)
        return self.net(combined)
```

```python
%%writefile /content/train.py

import os
import glob
import random
import numpy as np
import soundfile as sf
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
from model import Generator, Discriminator

BASE = "/content/voicebank_dataset"
CLEAN_DIR = os.path.join(BASE, "clean_trainset_28spk_wav")
NOISY_DIR = os.path.join(BASE, "noisy_trainset_28spk_wav")

SAVE_DIR = "/content/drive/MyDrive/GAN_Speech_Enhancement./checkpoints"
os.makedirs(SAVE_DIR, exist_ok=True)

SEGMENT_LENGTH = 16384
BATCH_SIZE = 8
EPOCHS = 10
LEARNING_RATE = 0.0002

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")


class SpeechDataset(Dataset):
    def __init__(self, clean_dir, noisy_dir):
        clean_files = {
            os.path.basename(p): p
            for p in glob.glob(os.path.join(clean_dir, "*.wav"))
        }
        noisy_files = {
            os.path.basename(p): p
            for p in glob.glob(os.path.join(noisy_dir, "*.wav"))
        }

        names = sorted(set(clean_files) & set(noisy_files))
        if not names:
            raise RuntimeError("No matching audio pairs found.")

        self.pairs = [(noisy_files[n], clean_files[n]) for n in names]
        print("Matched training pairs:", len(self.pairs))

    def __len__(self):
        return len(self.pairs)

    def __getitem__(self, index):
        noisy_path, clean_path = self.pairs[index]

        noisy, sr1 = sf.read(noisy_path, dtype="float32")
        clean, sr2 = sf.read(clean_path, dtype="float32")

        if noisy.ndim > 1:
            noisy = noisy.mean(axis=1)
        if clean.ndim > 1:
            clean = clean.mean(axis=1)

        if sr1 != 16000 or sr2 != 16000:
            raise ValueError(f"Expected 16000 Hz audio; got {sr1} and {sr2}")

        length = min(len(noisy), len(clean))
        noisy, clean = noisy[:length], clean[:length]

        if length >= SEGMENT_LENGTH:
            start = random.randint(0, length - SEGMENT_LENGTH)
            noisy = noisy[start:start + SEGMENT_LENGTH]
            clean = clean[start:start + SEGMENT_LENGTH]
        else:
            padding = SEGMENT_LENGTH - length
            noisy = np.pad(noisy, (0, padding))
            clean = np.pad(clean, (0, padding))

        scale = max(
            float(np.max(np.abs(noisy))),
            float(np.max(np.abs(clean))),
            1e-6
        )

        noisy = noisy / scale
        clean = clean / scale

        return (
            torch.from_numpy(noisy.copy()).unsqueeze(0),
            torch.from_numpy(clean.copy()).unsqueeze(0)
        )


def main():
    print("Training device:", DEVICE)

    dataset = SpeechDataset(CLEAN_DIR, NOISY_DIR)
    loader = DataLoader(
        dataset,
        batch_size=BATCH_SIZE,
        shuffle=True,
        num_workers=2,
        pin_memory=(DEVICE.type == "cuda"),
        drop_last=True
    )

    generator = Generator().to(DEVICE)
    discriminator = Discriminator().to(DEVICE)

    opt_g = optim.Adam(
        generator.parameters(), lr=LEARNING_RATE, betas=(0.5, 0.999)
    )
    opt_d = optim.Adam(
        discriminator.parameters(), lr=LEARNING_RATE, betas=(0.5, 0.999)
    )

    adv_loss = nn.BCEWithLogitsLoss()
    recon_loss = nn.L1Loss()

    for epoch in range(EPOCHS):
        generator.train()
        discriminator.train()
        total_g, total_d = 0.0, 0.0

        for step, (noisy, clean) in enumerate(loader, start=1):
            noisy = noisy.to(DEVICE)
            clean = clean.to(DEVICE)

            # Train discriminator
            with torch.no_grad():
                fake_detached = generator(noisy)

            opt_d.zero_grad(set_to_none=True)
            real_logits = discriminator(noisy, clean)
            fake_logits = discriminator(noisy, fake_detached)

            loss_d = 0.5 * (
                adv_loss(real_logits, torch.ones_like(real_logits))
                + adv_loss(fake_logits, torch.zeros_like(fake_logits))
            )
            loss_d.backward()
            opt_d.step()

            # Train generator
            opt_g.zero_grad(set_to_none=True)
            fake = generator(noisy)
            fake_logits = discriminator(noisy, fake)

            loss_g = (
                adv_loss(fake_logits, torch.ones_like(fake_logits))
                + 100.0 * recon_loss(fake, clean)
            )

            loss_g.backward()
            opt_g.step()

            total_g += loss_g.item()
            total_d += loss_d.item()

            if step % 100 == 0:
                print(
                    f"Epoch {epoch + 1}/{EPOCHS}, "
                    f"Step {step}/{len(loader)}, "
                    f"G={loss_g.item():.4f}, D={loss_d.item():.4f}"
                )

        avg_g = total_g / len(loader)
        avg_d = total_d / len(loader)

        checkpoint = {
            "epoch": epoch + 1,
            "generator_state_dict": generator.state_dict(),
            "discriminator_state_dict": discriminator.state_dict(),
            "optimizer_g_state_dict": opt_g.state_dict(),
            "optimizer_d_state_dict": opt_d.state_dict(),
            "generator_loss": avg_g,
            "discriminator_loss": avg_d
        }

        torch.save(
            checkpoint,
            os.path.join(SAVE_DIR, "generator_latest.pt")
        )
        torch.save(
            generator.state_dict(),
            os.path.join(SAVE_DIR, "generator_weights.pt")
        )

        print(
            f"Epoch {epoch + 1} complete: "
            f"G Loss={avg_g:.4f}, D Loss={avg_d:.4f}"
        )
        print("Checkpoint saved to Google Drive.")


if __name__ == "__main__":
    main()
```

```python
import sys
import numpy as np
import soundfile as sf
import torch
from pathlib import Path
from IPython.display import Audio, display

sys.path.insert(0, "/content")
from model import Generator

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

model_path = Path(
    "/content/drive/MyDrive/GAN_Speech_Enhancement./checkpoints/generator_weights.pt"
)

noisy_path = Path(
    "/content/voicebank_dataset/noisy_testset_wav"
)

clean_path = Path(
    "/content/voicebank_dataset/clean_testset_wav"
)

noisy_files = sorted(noisy_path.glob("*.wav"))
assert noisy_files, f"No noisy WAV files found in {noisy_path}"

noisy_file = noisy_files[0]
clean_file = clean_path / noisy_file.name

print("Test file:", noisy_file.name)
print("Clean reference exists:", clean_file.exists())

model = Generator().to(device)
weights = torch.load(model_path, map_location=device, weights_only=True)
model.load_state_dict(weights)
model.eval()

noisy, sr = sf.read(noisy_file, dtype="float32")
if noisy.ndim == 2:
    noisy = noisy.mean(axis=1)

if sr != 16000:
    import librosa
    noisy = librosa.resample(noisy, orig_sr=sr, target_sr=16000)

# Process audio in chunks of 16384 samples.
chunk_size = 16384
enhanced_parts = []

with torch.inference_mode():
    for start in range(0, len(noisy), chunk_size):
        chunk = noisy[start:start + chunk_size]
        original_length = len(chunk)

        if original_length < chunk_size:
            chunk = np.pad(chunk, (0, chunk_size - original_length))

        scale = max(float(np.max(np.abs(chunk))), 1e-6)
        x = torch.from_numpy((chunk / scale).copy()).float()
        x = x.unsqueeze(0).unsqueeze(0).to(device)

        y = model(x).squeeze().cpu().numpy()
        enhanced_parts.append(y[:original_length] * scale)

enhanced = np.concatenate(enhanced_parts)
enhanced = np.clip(np.nan_to_num(enhanced), -1, 1)

print("Original noisy audio:")
display(Audio(noisy, rate=16000))

print("GAN-enhanced audio:")
display(Audio(enhanced, rate=16000))

sf.write(
    "/content/enhanced_test.wav",
    enhanced,
    16000
)

print("Saved enhanced audio to /content/enhanced_test.wav")
```

```python
import os

PROJECT = "/content/drive/MyDrive/GAN_Speech_Enhancement."
os.makedirs(PROJECT, exist_ok=True)

model_code = '''
import torch
import torch.nn as nn

class Generator(nn.Module):
    def __init__(self):
        super().__init__()
        self.net = nn.Sequential(
            nn.Conv1d(1, 32, kernel_size=15, padding=7),
            nn.LeakyReLU(0.2),
            nn.Conv1d(32, 64, kernel_size=15, stride=2, padding=7),
            nn.BatchNorm1d(64),
            nn.LeakyReLU(0.2),
            nn.Conv1d(64, 128, kernel_size=15, stride=2, padding=7),
            nn.BatchNorm1d(128),
            nn.LeakyReLU(0.2),
            nn.ConvTranspose1d(128, 64, kernel_size=4, stride=2, padding=1),
            nn.BatchNorm1d(64),
            nn.ReLU(),
            nn.ConvTranspose1d(64, 32, kernel_size=4, stride=2, padding=1),
            nn.BatchNorm1d(32),
            nn.ReLU(),
            nn.Conv1d(32, 1, kernel_size=15, padding=7),
            nn.Tanh()
        )

    def forward(self, noisy_audio):
        enhanced = self.net(noisy_audio)
        if enhanced.shape[-1] > noisy_audio.shape[-1]:
            enhanced = enhanced[..., :noisy_audio.shape[-1]]
        elif enhanced.shape[-1] < noisy_audio.shape[-1]:
            enhanced = nn.functional.pad(
                enhanced,
                (0, noisy_audio.shape[-1] - enhanced.shape[-1])
            )
        return enhanced
'''

with open(os.path.join(PROJECT, "model.py"), "w") as f:
    f.write(model_code)

print("SUCCESS: model.py created in your existing project!")
```

```python
import os
import sys
import torch

PROJECT = "/content/drive/MyDrive/GAN_Speech_Enhancement."
sys.path.insert(0, PROJECT)

from model import Generator

checkpoint = os.path.join(PROJECT, "checkpoints", "generator_weights.pt")

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = Generator().to(device)

weights = torch.load(checkpoint, map_location=device, weights_only=True)

# Support either raw weights or a checkpoint dictionary
if isinstance(weights, dict) and "generator_state_dict" in weights:
    weights = weights["generator_state_dict"]
elif isinstance(weights, dict) and "state_dict" in weights:
    weights = weights["state_dict"]

model.load_state_dict(weights)
model.eval()

print("SUCCESS: Trained generator loaded!")
print("Device:", device)
print("Ready to test speech enhancement.")
```

```python
import glob
import numpy as np
import soundfile as sf
import librosa
from IPython.display import Audio, display

TEST_FOLDER = "/content/voicebank_dataset/noisy_testset_wav"

files = sorted(glob.glob(os.path.join(TEST_FOLDER, "*.wav")))

if not files:
    print("No test WAV files found. The dataset may need to be extracted first.")
else:
    input_file = files[0]
    audio, sr = sf.read(input_file)

    if audio.ndim > 1:
        audio = audio.mean(axis=1)

    if sr != 16000:
        audio = librosa.resample(
            audio.astype(np.float32),
            orig_sr=sr,
            target_sr=16000
        )

    audio = audio.astype(np.float32)
    audio = np.clip(audio, -1, 1)

    # Process in chunks to limit memory use
    chunk_size = 16384
    enhanced_parts = []

    with torch.no_grad():
        for start in range(0, len(audio), chunk_size):
            chunk = audio[start:start + chunk_size]
            scale = max(float(np.max(np.abs(chunk))), 1e-7)
            chunk_tensor = torch.from_numpy(
                (chunk / scale).copy()
            ).view(1, 1, -1).to(device)

            result = model(chunk_tensor)
            result = result.squeeze().cpu().numpy() * scale
            enhanced_parts.append(result)

    enhanced_audio = np.concatenate(enhanced_parts)
    enhanced_audio = np.clip(enhanced_audio, -1, 1)

    sf.write("/content/enhanced_test.wav", enhanced_audio, 16000)

    print("Original noisy audio:")
    display(Audio(audio, rate=16000))

    print("Enhanced audio:")
    display(Audio(enhanced_audio, rate=16000))

    print("Enhanced file saved to /content/enhanced_test.wav")
```

```python
import os

PROJECT = "/content/drive/MyDrive/GAN_Speech_Enhancement."

app_code = r'''
import streamlit as st
import torch
import numpy as np
import soundfile as sf
import librosa
import io
from pathlib import Path
from model import Generator

st.set_page_config(
    page_title="SonicShield AI",
    page_icon="🎙️",
    layout="wide"
)

PROJECT = Path(__file__).parent
CHECKPOINT = PROJECT / "checkpoints" / "generator_weights.pt"

@st.cache_resource
def load_model():
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    model = Generator().to(device)

    weights = torch.load(
        CHECKPOINT, map_location=device, weights_only=True
    )

    if isinstance(weights, dict) and "generator_state_dict" in weights:
        weights = weights["generator_state_dict"]
    elif isinstance(weights, dict) and "state_dict" in weights:
        weights = weights["state_dict"]

    model.load_state_dict(weights)
    model.eval()
    return model, device

def enhance_audio(audio, model, device):
    audio = np.asarray(audio, dtype=np.float32)
    audio = np.nan_to_num(audio)

    enhanced_parts = []
    chunk_size = 16384

    with torch.no_grad():
        for start in range(0, len(audio), chunk_size):
            chunk = audio[start:start + chunk_size]
            scale = max(float(np.max(np.abs(chunk))), 1e-7)

            tensor = torch.from_numpy(
                (chunk / scale).copy()
            ).view(1, 1, -1).to(device)

            result = model(tensor)
            result = result.squeeze().cpu().numpy() * scale
            enhanced_parts.append(result)

    enhanced = np.concatenate(enhanced_parts)
    return np.clip(enhanced, -1, 1)

st.title("🎙️ SonicShield AI")
st.subheader("GAN-Based Speech Enhancement Studio")
st.write(
    "Upload a noisy recording, process it with your trained GAN generator, "
    "and compare the input and enhanced speech."
)

if not CHECKPOINT.exists():
    st.error(f"Trained model not found: {CHECKPOINT}")
    st.stop()

try:
    model, device = load_model()
    st.success(f"Trained GAN model loaded successfully • Device: {device}")
except Exception as e:
    st.error(f"Could not load model: {e}")
    st.stop()

uploaded = st.file_uploader(
    "Upload a noisy audio recording",
    type=["wav", "flac", "ogg"]
)

if uploaded is not None:
    try:
        raw_audio, sample_rate = sf.read(
            io.BytesIO(uploaded.getvalue()),
            dtype="float32"
        )

        if raw_audio.ndim == 2:
            raw_audio = raw_audio.mean(axis=1)

        if len(raw_audio) == 0:
            st.error("The uploaded audio is empty.")
            st.stop()

        st.write(f"Original sample rate: {sample_rate} Hz")
        st.audio(uploaded.getvalue())

        audio_16k = raw_audio
        if sample_rate != 16000:
            audio_16k = librosa.resample(
                raw_audio,
                orig_sr=sample_rate,
                target_sr=16000
            )

        st.info("Ready to process the recording.")

        if st.button("✨ Enhance Speech", type="primary"):
            with st.spinner("Enhancing speech..."):
                enhanced = enhance_audio(audio_16k, model, device)

            buffer = io.BytesIO()
            sf.write(buffer, enhanced, 16000, format="WAV")
            wav_bytes = buffer.getvalue()

            st.success("Enhancement finished!")
            left, right = st.columns(2)

            with left:
                st.markdown("### Original recording")
                st.audio(uploaded.getvalue())

            with right:
                st.markdown("### Enhanced recording")
                st.audio(wav_bytes, format="audio/wav")

            st.download_button(
                "⬇️ Download Enhanced WAV",
                data=wav_bytes,
                file_name="enhanced_speech.wav",
                mime="audio/wav"
            )

            st.markdown("### Audio measurements")
            c1, c2, c3 = st.columns(3)
            c1.metric("Duration", f"{len(audio_16k)/16000:.2f} sec")
            c2.metric("Input sample rate", f"{sample_rate} Hz")
            c3.metric("Output sample rate", "16000 Hz")

            st.caption(
                "This is an experimental GAN model. Listen carefully to "
                "confirm that speech quality improved; enhancement can "
                "sometimes introduce distortion."
            )

    except Exception as e:
        st.error(f"Could not process this recording: {e}")

st.divider()
st.caption("GAN Speech Enhancement • Academic Project")
'''

with open(os.path.join(PROJECT, "app.py"), "w", encoding="utf-8") as f:
    f.write(app_code)

print("SUCCESS: app.py created in your existing Google Drive project!")
print("Location:", os.path.join(PROJECT, "app.py"))
```
